```python
# Step 1: Import Required Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
# Step 2: Create Movie Dataset

data = {
    "Movie": [
        "Leo", "Vikram", "Jailer", "Master", "Beast",
        "Kaithi", "Doctor", "Don", "Varisu", "Thunivu",
        "Jai Bhim", "Soorarai Pottru", "96", "Vada Chennai", "Asuran",
        "Premam", "Kumbalangi Nights", "KGF", "RRR", "Pushpa"
    ],

    "Genre": [
        "Action", "Action", "Action", "Action", "Action",
        "Action", "Comedy", "Comedy", "Drama", "Action",
        "Drama", "Drama", "Romance", "Crime", "Drama",
        "Romance", "Drama", "Action", "Action", "Action"
    ],

    "Rating": [
        8.0, 8.3, 7.9, 7.3, 5.2,
        8.5, 7.3, 7.0, 5.6, 6.1,
        8.7, 8.7, 8.6, 8.6, 8.4,
        8.3, 8.5, 8.2, 8.0, 7.6
    ],

    "Votes": [
        15000, 18000, 14000, 12000, 10000,
        13000, 11000, 9000, 8000, 8500,
        16000, 15000, 14000, 13000, 12000,
        10000, 9000, 20000, 18000, 17000
    ],

    "Audience_Preference": [
        "High", "High", "High", "Medium", "Low",
        "High", "Medium", "Medium", "Low", "Medium",
        "High", "High", "High", "High", "High",
        "High", "High", "High", "High", "High"
    ]
}

df = pd.DataFrame(data)

print("Dataset created successfully!")
df.head()
```

```python
# Step 3: Display Complete Dataset

df
```

```python
# Step 4: Dataset Information

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
# Step 5: Check Missing Values

print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
# Step 6: Statistical Summary

df.describe()
```

```python
# Step 7: Top Rated Movies

top_movies = df.sort_values(by="Rating", ascending=False)

top_movies[["Movie", "Genre", "Rating"]].head(10)
```

```python
# Step 8: Average Rating by Genre

genre_rating = df.groupby("Genre")["Rating"].mean().sort_values(ascending=False)

print(genre_rating)
```

```python
# Step 10: Rating Distribution

plt.figure(figsize=(8,5))

plt.hist(df["Rating"], bins=6, edgecolor="black")

plt.title("Distribution of Movie Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Movies")

plt.show()
```

```python
# Step 11: Genre-wise Average Rating

plt.figure(figsize=(8,5))

genre_rating.plot(kind="bar")

plt.title("Average Movie Rating by Genre")
plt.xlabel("Genre")
plt.ylabel("Average Rating")

plt.xticks(rotation=0)

plt.show()
```

```python
# Step 12: Top 10 Rated Movies

top_10 = df.sort_values(by="Rating", ascending=False).head(10)

plt.figure(figsize=(10,5))

plt.bar(top_10["Movie"], top_10["Rating"])

plt.title("Top 10 Rated Movies")
plt.xlabel("Movie")
plt.ylabel("Rating")

plt.xticks(rotation=45)

plt.show()
```

```python
# Step 13: Rating vs Number of Votes

plt.figure(figsize=(8,5))

plt.scatter(df["Votes"], df["Rating"])

plt.title("Movie Ratings vs Number of Votes")
plt.xlabel("Number of Votes")
plt.ylabel("Rating")

plt.show()
```

```python
# Step 14: Audience Preference Analysis

preference_count = df["Audience_Preference"].value_counts()

print(preference_count)
```

```python
plt.figure(figsize=(7,5))

preference_count.plot(kind="bar")

plt.title("Audience Preference Distribution")
plt.xlabel("Audience Preference")
plt.ylabel("Number of Movies")

plt.xticks(rotation=0)

plt.show()
```

```python
# Step 15: Final Analysis

highest_rated_movie = df.loc[df["Rating"].idxmax(), "Movie"]
highest_rating = df["Rating"].max()

best_genre = genre_rating.idxmax()
best_genre_rating = genre_rating.max()

most_preferred = preference_count.idxmax()

print("===== MOVIE RATINGS AND AUDIENCE PREFERENCE ANALYSIS =====")

print("\nHighest Rated Movie:")
print(highest_rated_movie, "-", highest_rating)

print("\nBest Rated Genre:")
print(best_genre, "-", round(best_genre_rating, 2))

print("\nMost Common Audience Preference:")
print(most_preferred)

print("\nTotal Movies Analysed:")
print(len(df))
```
